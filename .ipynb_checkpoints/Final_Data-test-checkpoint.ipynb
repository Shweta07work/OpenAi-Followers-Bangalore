# Analyzing the **Final Collected Data**

The final dataset is analyzed using the collected data instead of the **demo file** used earlier.

The analysis focuses on identifying users with thMaximumhest posts, followers, and following**, along with different **account categories** and their user counts.

```python
with open("finaldata.txt", encoding='utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
# Parsing the data in pure python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split('\n')
    username = sep_chunk[0]
    no_of_posts = int(sep_chunk[1].split(" post")[0].replace(",",""))
    no_of_followers = float(sep_chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M", ""))
    if("K" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000)
    elif("M" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000000)
    else:
        no_of_followers = int(no_of_followers)


    
    no_of_following = float(sep_chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M", ""))
    if("K" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000)
    elif("M" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000000)
    else:
        no_of_following = int(no_of_following)

    
    name = sep_chunk[4]
    if(len(sep_chunk)> 5):
        type_of_categories = sep_chunk[5]
        bio = "\n".join(sep_chunk[6:])
    else:
        type_of_categories = "Unknown"
        bio = ""
    return{"username": username, "no_of_posts": no_of_posts, "no_of_followers": no_of_followers, "type_of_categories": type_of_categories, 
          "no_of_following": no_of_following, "name": name, "type_of_categories": type_of_categories, "bio": bio}
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s = json.dumps(all_chunks, indent=4)
with open("data.json","w") as f:
    f.write(s)
```

## Who has the maximum post?
- Code to find maximum posts.

```python
max = 0
for chunk in all_chunks:
    if(max<chunk['no_of_posts']):
        max = chunk['no_of_posts']
        chunk_with_max_post = chunk
print(chunk_with_max_post['username'])
```

## Who has the maximum followers?
- Code to find who has maximum followers.

```python
max = 0
for chunk in all_chunks:
    if(max<chunk['no_of_followers']):
        max = chunk['no_of_followers']
        chunk_with_max_followers = chunk
print(chunk_with_max_followers['username'])
```

## Who follows maximum people?
- Code to find who follows maximum people. 

```python
max = 0
for chunk in all_chunks:
    if(max<chunk['no_of_following']):
        max = chunk['no_of_following']
        chunk_with_max_followings = chunk
print(chunk_with_max_followings['username'])
```

## How many categories?
- Code to find how many categories. 

```python
categories = set()
for chunk in all_chunks:
    categories.add(chunk['type_of_categories'])
print(f"There are {len(categories)} categories")
```
